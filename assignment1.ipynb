```python
from sklearn.ensemble import IsolationForest
import matplotlib.pyplot as plt
import numpy as np
cpu_usage = [
    32, 35, 34, 36, 33,
    35, 37, 34, 36, 35,
    34, 33, 36, 35, 34,
    92, 35, 33, 95, 34
]
```

```python
X = np.array(cpu_usage).reshape(-1, 1)
```

```python
model = IsolationForest(contamination=0.10, random_state=42)
```

```python
predictions = model.fit_predict(X)
```

```python
print("Detected Anomalies:")
for usage, pred in zip(cpu_usage, predictions):
    if pred == -1:
        print(f"Anomaly: {usage}% CPU")
```

```python
plt.figure(figsize=(10, 5))
colors = ['red' if p == -1 else 'blue' for p in predictions]
intervals = range(1, len(cpu_usage) + 1)
```

```python
plt.scatter(intervals, cpu_usage, color=colors, s=100)
plt.plot(intervals, cpu_usage, color='gray', linestyle='--', zorder=1)
plt.title('AIOps Server CPU Utilization Monitoring')
plt.xlabel('Time Interval')
plt.ylabel('CPU Usage (%)')
plt.xticks(intervals)
plt.grid(True, alpha=0.3)
plt.show()
```
